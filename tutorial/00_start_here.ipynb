The model file is `mano_v1_2/models/MANO_RIGHT.pkl` in the parent directory. The environment is Python 3.11. Dependencies are listed in `MANO/requirements.txt`.

Paper: Romero, Tzionas, Black, *Embodied Hands: Modeling and Capturing Hands and Bodies Together*, SIGGRAPH Asia 2017. MANO is the hand formulation of SMPL. The release contains a right hand and a left hand, each with 778 vertices, 1538 faces, and 16 joints. Units are metres.

## What you should be able to do at the end

1. Write the three pose indices of any joint in the 48-D pose.
2. Read an axis-angle as an axis plus an angle in radians, and know that the four fingers of this official right hand flex toward the palm about local +Z.
3. Split one forward pass into shape, joint regression, pose blend, forward kinematics, and skinning, and name the array each step reads.
4. Load `MANO_RIGHT.pkl`, flex a chosen finger, change `betas`, and read out 16 joints and 5 fingertips.
5. Tell axis-angle zeros apart from PCA-coefficient zeros.
6. Know that fingertips are vertices, that the left hand comes from `MANO_LEFT.pkl`, and why MANO's 16 joints cannot be subtracted from MSRA's 21 points in index order.
